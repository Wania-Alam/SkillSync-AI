```python
# Cleaning The Data
```

```python
# Make a copy so raw data stays untouched
skills_clean = skills_df.copy()

def clean_skill(skill):
    skill = str(skill)

    # Convert to lowercase
    skill = skill.lower()

    # Remove newline, tab and extra whitespace
    skill = skill.replace("\n", " ")
    skill = skill.replace("\t", " ")
    skill = re.sub(r"\s+", " ", skill)

    # Remove spaces at beginning/end
    skill = skill.strip()

    return skill

skills_clean["clean_skill"] = skills_clean["Skill"].apply(clean_skill)

skills_clean.head(20)
```

```python
print("Rows:", len(skills_clean))
print("Unique cleaned skills:", skills_clean["clean_skill"].nunique())
print("Duplicates after cleaning:",
      skills_clean["clean_skill"].duplicated().sum())

print("Empty after cleaning:",
      (skills_clean["clean_skill"] == "").sum())
```

```python
skills_clean = skills_clean.drop_duplicates(
    subset="clean_skill"
).reset_index(drop=True)
```

```python
# Make a copy before filtering
skills_filtered = skills_clean.copy()

# Known short technical/professional skills
protected_short_skills = {
    "c",
    "r",
    "c++",
    "c#",
    "f#",
    "g++",
    "a++",
    "api",
    "aws",
    "sql",
    "git",
    "npm",
    "css",
    "php",
    "vue",
    "xml",
    "ios",
    "ios",
    "ml",
    "ai",
    "etl",
    "nlp",
    "mvc",
    "uml",
    "ssh",
    "ftp",
    "pdf",
    "svg",
    "html",
    "css",
    "js",
    "go",
    "os",
    "ui",
    "ux",
    "it",
    "cad",
    "gis",
    "erp",
    "crm",
    "sap",
    "sdk",
    "dns",
    "tcp",
    "udp",
    "ram",
    "rom",
    "cpu",
    "gpu",
    "x86",
    "3d"
}

def is_valid_short_skill(skill):
    """
    Keep short skills if:
    1. They have more than 3 characters, OR
    2. They are explicitly present in our protected list.
    """
    
    if len(skill) > 3:
        return True
    
    if skill in protected_short_skills:
        return True
    
    return False


skills_filtered = skills_filtered[
    skills_filtered["clean_skill"].apply(is_valid_short_skill)
].reset_index(drop=True)

print("Original skills:", len(skills_clean))
print("After short-skill filtering:", len(skills_filtered))
print("Removed:", len(skills_clean) - len(skills_filtered))
```

```python
short_remaining = skills_filtered[
    skills_filtered["skill_length"] <= 3
]["clean_skill"]

print("Remaining short skills:")
print(short_remaining.to_list())
```

```python
skills_filtered["skill_length"] = (
    skills_filtered["clean_skill"].str.len()
)

skills_filtered["word_count"] = (
    skills_filtered["clean_skill"]
    .str.split()
    .str.len()
)
```
